## Stap 2: Het Stappenplan per Scenario

### 📘 Hoofdstuk 3: Eén steekproef testen (1 Variabele)
Je vergelijkt de data van één groep met een vooraf bepaalde norm, een vast getal of een landelijk gemiddelde.

* **Vraag:** "Is het gemiddelde significant anders dan X?" of "Is de kans groter dan Y?"
* **Voorbeeld:** Je test een nieuw OCR-script en wilt weten: *Is de gemiddelde framerate-drop van dit script in een testomgeving minder dan 5%?*
* **Welke test?**
    * Bij een gemiddelde (getallen): **One-sample t-test** (`scipy.stats.ttest_1samp`).
    * Bij een percentage/proportie (Ja/Nee): **One-sample z-test** (`statsmodels.stats.proportion.proportions_ztest`).

### 📙 Hoofdstuk 4: Categorieën vergelijken (Kwalitatief vs Kwalitatief)
Je hebt data die in "bakjes" of categorieën valt en je wilt weten of de verdeling toeval is, of dat er een verband is. Er komen hier (bijna) géén gemiddeldes aan te pas, je telt **frequenties**.

* **Vraag:** "Is er een verband tussen..." of "Zijn de variabelen onafhankelijk?"
* **Voorbeeld:** Je analyseert de effectiviteit van een chatfilter op een gaming platform. *Is er een verband tussen het type leetspeak (cijfers vs speciale symbolen) en of het bericht geblokkeerd wordt (Ja/Nee)?*
* **Welke test?**
    * Verband tussen 2 categorieën: **Chi-kwadraat test voor onafhankelijkheid** (`scipy.stats.chi2_contingency`).
    * Kijken of 1 categorie voldoet aan een verwachte verdeling: **Goodness of fit test** (`scipy.stats.chisquare`).
    * *Visualisatie:* Mosaic plot (`statsmodels.graphics.mosaicplot.mosaic`).

### 📗 Hoofdstuk 5: Groepen vergelijken (Kwantitatief vs Kwalitatief)
Je hebt een "score" of "tijd" (getal) en je wilt die vergelijken tussen twee of meerdere groepen (categorie). 

* **Vraag:** "Scoort groep A beter dan groep B?" of "Is er een verschil voor en na de ingreep?"
* **Voorbeeld:** *Is er een verschil in de rekentijd (getal) tussen Moderatie-Algoritme A en Moderatie-Algoritme B (2 groepen)?*
* **Welke test?**
    * 2 onafhankelijke groepen (losse datasets): **Independent T-test** (`scipy.stats.ttest_ind`).
    * 2 afhankelijke metingen (bijv. zelfde dataset, meting vóór en meting ná): **Paired T-test** (`scipy.stats.ttest_rel`).
    * Meer dan 2 groepen (Algoritme A, B én C): **ANOVA** (`scipy.stats.f_oneway`).
    * *Visualisatie:* Boxplot of Barplot met error bars.

### 📕 Hoofdstuk 6: Voorspellen en Verbanden (Kwantitatief vs Kwantitatief)
Beide variabelen zijn continue getallen. Je wilt weten of de ene oploopt als de andere oploopt, of je wilt een voorspelling doen.

* **Vraag:** "Is er een lineair verband?" of "Wat is de verwachte Y als X toeneemt?"
* **Voorbeeld:** *Is er een lineair verband tussen de lengte van een chatbericht (aantal karakters) en de scantijd van het algoritme (in milliseconden)?*
* **Welke test?**
    * Om het verband te meten: **Correlatie** (bijv. Pearson).
    * Om een model/voorspelling te maken: **Lineaire Regressie** (`sklearn.linear_model.LinearRegression`).
    * *Visualisatie:* Scatterplot met een trendlijn (`sns.regplot`).

### 📓 Hoofdstuk 7: Tijdreeksen (Time Series)
Dit is makkelijk te herkennen: je X-as is vrijwel altijd **Tijd** (jaren, maanden, dagen). Je kijkt naar trends, seizoenspatronen of doet voorspellingen (forecasting) voor de toekomst.

* **Vraag:** "Wat is de verwachte trend voor het komende jaar?"
* **Voorbeeld:** *Het voorspellen van het aantal actieve gebruikers op een server per maand, gebaseerd op de data van de afgelopen drie jaar.*
* **Welke test/model?**
    * Vlakken van data: **Moving Average (SMA)** of **Exponential Smoothing (EMA)** (`statsmodels.tsa.holtwinters.SimpleExpSmoothing`).
    * Data met een trend (stijgend/dalend): **Holt's method** (`statsmodels.tsa.holtwinters.Holt`).
    * Data met een trend én seizoen (bijv. elke zomer drukker): **Holt-Winters** (`statsmodels.tsa.holtwinters.ExponentialSmoothing`).
    * *Visualisatie:* Line plot over tijd.
