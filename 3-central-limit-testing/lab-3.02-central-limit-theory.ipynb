## Exercise 1 - Cholesterol 

A health survey between 1988 and 1994 indicated that the mean cholesterol value for women between the age of 20 and 29 was 183 mg/dl, with a standard deviation equal to 36. We now take a random sample of 81 women.

1. Plot the probability distribution of the sample mean $\overline{x}$.
2. What is the probability of the sample mean being $\overline{x} < 185$? (Answer ≈ 69,1%)
3. What is the probability of $175 < \overline{x} < 185$ (Answer ≈ 66,9%)
4. What is the probability of $\overline{x} > 190$ (Answer ≈ 4,0%)

```python
pop_mean = 183
pop_std = 36
n = 81
stderr = pop_std / np.sqrt(n)

xvals = np.linspace(pop_mean - 4 * stderr, pop_mean + 4 * stderr, 201)
yvals = stats.norm.pdf(xvals, loc=pop_mean, scale=stderr)
plt.plot(xvals, yvals)
plt.axvline(pop_mean, color='red', linestyle='--')
print(f"P(M < 185) = {stats.norm.cdf(185, loc=pop_mean, scale=stderr):.4f}")
print(f"P(175 < M < 185) = {stats.norm.cdf(185, loc=pop_mean, scale=stderr) - stats.norm.cdf(175, loc=pop_mean, scale=stderr):.4f}")
print(f"P(M > 190) = {1 - stats.norm.cdf(190, loc=pop_mean, scale=stderr):.4f}")

plt.axvline(175, color='orange', linestyle='--')
plt.axvline(185, color='blue', linestyle='--')
plt.axvline(190, color='green', linestyle='--')
```
