## Exercise 1 - rlanders

Load the dataset data/rlanders.csv. Use the column Money.  
We assume that the values from this sample are normally distributed around a population mean $\mu$ that is unknown, but that the standard deviation of the population is known, i.e. $\sigma = 98$.


```python
df = pd.read_csv('https://raw.githubusercontent.com/HoGentTIN/dsai-en-labs/main/data/rlanders.csv').set_index(['ID'])
# Show the first few records of the Data Frame
df.head()
```

Calculate a 99% confidence interval for the population mean (Answer: [484.191, 516.121]).

```python
# 99% CI for population mean, with known sigma = 98
money = df["Money"].dropna()
n = len(money)
xbar = money.mean()
sigma = 98

alpha = 0.01
z = stats.norm.ppf(1 - alpha / 2)
me = z * sigma / np.sqrt(n)

ci_99_known = (xbar - me, xbar + me)
print(f"99% CI (sigma known): [{ci_99_known[0]:.3f}, {ci_99_known[1]:.3f}]")
```

Calculate a 95% confidence interval for the population mean. (Answer =  [488.008, 512.304])

```python
# 95% CI for population mean, with known sigma = 98
alpha = 0.05
z = stats.norm.ppf(1 - alpha / 2)
me = z * sigma / np.sqrt(n)

ci_95_known = (xbar - me, xbar + me)
print(f"95% CI (sigma known): [{ci_95_known[0]:.3f}, {ci_95_known[1]:.3f}]")
```

Suppose σ is unknown, calculate a 95% confidence interval for the population mean. (Answer = [487.319, 512.993])

```python
alpha = 0.05 
s = money.std(ddof=1)
t_crit = stats.t.ppf(1 - alpha / 2, df=n-1)
me = t_crit * s / np.sqrt(n)

ci_95_unknown = (xbar - me, xbar + me)
print(f"95% CI (unknown sigma): [{ci_95_unknown[0]:.3f}, {ci_95_unknown[1]:.3f}]")
```

Finally, suppose the sample consists of only the first 25 observations in this dataset. Calculate a 95% confidence interval for this situation.  
(Answer = [450.291,536.669])

```python
# 95% CI using only the first 25 observations (sigma unknown -> t interval)
money25 = df["Money"].dropna().iloc[:25]
n25 = len(money25)
xbar25 = money25.mean()
s25 = money25.std(ddof=1)

alpha = 0.05
t_crit25 = stats.t.ppf(1 - alpha / 2, df=n25 - 1)
me25 = t_crit25 * s25 / np.sqrt(n25)

ci_95_first25 = (xbar25 - me25, xbar25 + me25)
print(f"95% CI (first 25, sigma unknown): [{ci_95_first25[0]:.3f}, {ci_95_first25[1]:.3f}]")
```

```python
print(stats.norm.ppf(1 - 0.05 / 2))
print(stats.norm.ppf(1 - 0.01 / 2))
```

```python
# Exercise 2 - Confidence Interval Questions

# 1. How do you calculate the upper and lower limits of a 95% and 99% confidence interval?
print("1. Confidence Interval Calculation:")
print("   For KNOWN sigma (z-interval):")
print("   Lower limit = x̄ - z*(σ/√n)")
print("   Upper limit = x̄ + z*(σ/√n)")
print("   where z is the critical value from the standard normal distribution")
print()
print("   For UNKNOWN sigma (t-interval):")
print("   Lower limit = x̄ - t*(s/√n)")
print("   Upper limit = x̄ + t*(s/√n)")
print("   where t is the critical value from the t-distribution")
print()

# Example with our data:
alpha_95 = 0.05
alpha_99 = 0.01
z_95 = stats.norm.ppf(1 - alpha_95 / 2)
z_99 = stats.norm.ppf(1 - alpha_99 / 2)

print(f"   95% CI: z-critical value = {z_95:.4f}")
print(f"   99% CI: z-critical value = {z_99:.4f}")
print()

# 2. A 99% confidence interval is [wider/narrower/the same width] as a 95% CI. Why?
print("2. Width Comparison:")
print("   A 99% CI is WIDER than a 95% CI.")
print("   Why? Because a higher confidence level requires a larger critical value:")
print(f"   - 95% CI uses z = {z_95:.4f}")
print(f"   - 99% CI uses z = {z_99:.4f}")
print("   Larger critical value → larger margin of error → wider interval")
print()

# 3. What would a 100% confidence interval look like?
print("3. 100% Confidence Interval:")
print("   A 100% CI would have limits from -∞ to +∞")
print("   Or more practically: (-∞, +∞)")
print("   Why? The critical value at 100% confidence would be infinite.")
print("   This is impractical and tells us nothing useful about the population mean.")
print("   This is why confidence levels are typically 90%, 95%, or 99%.")
```
