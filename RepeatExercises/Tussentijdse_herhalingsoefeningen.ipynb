Een producent van sinaasappelsap koopt al zijn sinaasappels bij één grote sinaasappelproducent. De hoeveelheid sap die uit elk van deze sinaasappels wordt geperst is ongeveer normaal verdeeld, met een gemiddelde van 133 gram en een standaardafwijking van 11,3 gram.

1. Wat is de kans dat een willekeurig gekozen sinaasappel tussen de 133 gram en 142 gram sap bevat?
2. 77% van de sinaasappels zal minstens hoeveel gram sap bevatten?
3. 80% van de sinaasappels bevat een hoeveelheid sap die tussen welke twee waarden (symmetrisch verdeeld) rond het gemiddelde ligt?

```python
mu1, sigma1 = 133, 11.3
# 1. Kans tussen 133 en 142
prob1 = stats.norm.cdf(142, mu1, sigma1) - stats.norm.cdf(133, mu1, sigma1)
# 2. 77% minstens x gram (dit is het 23ste percentiel: 1 - 0.77 = 0.23)
val2 = stats.norm.ppf(1 - 0.77, mu1, sigma1)
# 3. 80% symmetrisch rond gemiddelde (bovenste 10% en onderste 10%)
lower3 = stats.norm.ppf(0.10, mu1, sigma1)
upper3 = stats.norm.ppf(0.90, mu1, sigma1)

print(f"Sinaasappelsap: 1) {prob1:.4f}, 2) {val2:.2f} gram, 3) tussen {lower3:.2f} en {upper3:.2f} gram")
```

Een artikel rapporteerde dat voor een steekproef van 50 keukens met gaskooktoestellen die gedurende een ganse week werden gemonitord, het gemiddelde CO2-niveau in de steekproef 654.16 ppm was en de standaardafwijking 164.43 ppm.

Bereken en interpreteer een 95% confidence interval voor het ware gemiddelde CO2 -niveau in de huizen die betrokken waren in de steekproef.

```python
n = 50
mean_sample = 654.16
std_sample = 164.43
# 95% CI voor gemiddelde (t-distributie omdat populatie-std onbekend is)
ci = stats.t.interval(0.95, df=n-1, loc=mean_sample, scale=std_sample/np.sqrt(n))
print(f"CO2 CI: Het 95% betrouwbaarheidsinterval is {ci}")
```

Jan is lid van een schietclub. De kans dat hij het target raakt, is 25%. Dit is onafhankelijk van het feit of hij de vorige keer het target al dan niet geraakt heeft. Hij vuurt vier keer af.
1. Wat is de kans dat Jan ten minste 1 keer het target raakt?
2. Wat is de kans dat Jan exact 2 keer het target raakt?
3. Wat is de kans dat Jan ten minste 3 keer het target raakt?

```python
n_shots = 4
p_hit = 0.25
# 1. Ten minste 1 keer (1 - P(0))
prob_at_least_1 = 1 - stats.binom.pmf(0, n_shots, p_hit)
# 2. Exact 2 keer
prob_exact_2 = stats.binom.pmf(2, n_shots, p_hit)
# 3. Ten minste 3 keer (P(3) + P(4))
prob_at_least_3 = stats.binom.pmf(3, n_shots, p_hit) + stats.binom.pmf(4, n_shots, p_hit)

print(f"Schietclub: 1) {prob_at_least_1:.4f}, 2) {prob_exact_2:.4f}, 3) {prob_at_least_3:.4f}")
```

De vulhoeveelheid van flessen frisdrank is normaal verdeeld, met een gemiddelde van 2,0 liter en een standaardafwijking van 0,05 liter. Als flessen minder dan 95% van de vermelde netto-inhoud bevatten (in dit geval 1,90 liter), kan de fabrikant een boete krijgen . Flessen met een netto inhoud van meer dan 2,10 liter kunnen overmatig morsen bij het openen. Welk deel van de flessen bevat

1. tussen 1,90 en 2,0 liter?
2. minder dan 1,90 liter of meer dan 2,10 liter?
3. 99% van de flessen bevat minstens hoeveel frisdrank?

```python
mu2, sigma2 = 2.0, 0.05
# 1. Tussen 1.90 en 2.0
prob_f1 = stats.norm.cdf(2.0, mu2, sigma2) - stats.norm.cdf(1.90, mu2, sigma2)
# 2. < 1.90 of > 2.10
prob_f2 = stats.norm.cdf(1.90, mu2, sigma2) + (1 - stats.norm.cdf(2.10, mu2, sigma2))
# 3. 99% heeft minstens x (1ste percentiel)
val_f3 = stats.norm.ppf(0.01, mu2, sigma2)

print(f"Frisdrank: 1) {prob_f1:.4f}, 2) {prob_f2:.4f}, 3) {val_f3:.4f} liter")
```
